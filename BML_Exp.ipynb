```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, classification_report, silhouette_score

# 1. Extract data from UCI ML repository
data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columns = ["sepal_length", "sepal_width", "petal_length", "petal_width", "class"]
df = pd.read_csv(data_url, names=columns)

# Convert categorical class labels to numerical values
df['class'] = df['class'].astype('category').cat.codes

# 2. Data Preprocessing
X = df.drop(columns=['class'])
y = df['class']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 3. Logistic Regression
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)
y_pred_log = log_reg.predict(X_test)
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_log))

# 4. Naïve Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred_nb = nb.predict(X_test)
print("Naïve Bayes Accuracy:", accuracy_score(y_test, y_pred_nb))

# 5. Decision Tree
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))

# 6. Random Forest
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))

# 7. KNN Classifier
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
print("KNN Classifier Accuracy:", accuracy_score(y_test, y_pred_knn))

# 8. K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_train)
kmeans_pred = kmeans.predict(X_test)
print("K-Means Clustering Labels:", np.unique(kmeans_pred))

# 9. Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("PCA Explained Variance Ratio:", pca.explained_variance_ratio_)

# 10. Random Forest Ensemble Method
rf_ensemble = RandomForestClassifier(n_estimators=100)
rf_ensemble.fit(X_train, y_train)
y_pred_rf_ensemble = rf_ensemble.predict(X_test)
print("Random Forest Ensemble Accuracy:", accuracy_score(y_test, y_pred_rf_ensemble))

# 11. Boosting Ensemble Method
boosting = GradientBoostingClassifier()
boosting.fit(X_train, y_train)
y_pred_boost = boosting.predict(X_test)
print("Boosting Ensemble Accuracy:", accuracy_score(y_test, y_pred_boost))

# 12. Evaluate Performance Metrics
print("Classification Report for Supervised Models:")
print(classification_report(y_test, y_pred_rf))
```
